### Simple Create agent with model (no harness)

```python
from langchain.agents import create_agent

agent = create_agent(
    model="openai:gpt-5.5"
)

# res = agent.invoke(
#     {
#         "messages": [
#             {
#                 "role" : "user",
#                 "content" : "I want to go to UK from INDIA"
#             }
#         ]
#     }
# )

res = agent.invoke(
    {
        "messages": [
            {
                "role" : "user",
                "content" : "For the work purpose"
            }
        ]
    }
)

print(res['messages'][-1].content)

```
