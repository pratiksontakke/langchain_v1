### Creat agent with msg history

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver


agent = create_agent(
    model="openai:gpt-5.5",
    checkpointer=InMemorySaver() # 1
)

config1 = {"configurable": {"thread_id": "pratik-session-01"}} # 2
 
res1 = agent.invoke(
    {"messages": [{"role": "user", "content": "I want to go UK from INDIA"}]},
    config=config1 # 3
)

print("res1")
print(res1["messages"][-1].content)


res2 = agent.invoke(
    {"messages": [{"role": "user", "content": "For work purpose"}]},
    config=config1 # 3
) 

print("\n\n res2")
print(res2["messages"][-1].content)



res3 = agent.invoke(
    {"messages": [{"role": "user", "content": "summarise my history "}]},
    config=config1 # 3
) 

print("\n\n res3")
print(res3["messages"][-1].content)

```

### Create agent with msg history with system prompt

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

system_prompt = "Always reply in one line"

agent = create_agent(
    model="openai:gpt-5.5",
    checkpointer=InMemorySaver(),
    system_prompt=system_prompt
)

config1 = {"configurable": {"thread_id": "pratik-session-01"}} # 2
 
res1 = agent.invoke(
    {"messages": [{"role": "user", "content": "I want to go UK from INDIA"}]},
    config=config1 # 3
)

print("res1")
print(res1["messages"][-1].content)


res2 = agent.invoke(
    {"messages": [{"role": "user", "content": "For work purpose"}]},
    config=config1 # 3
) 

print("\n\n res2")
print(res2["messages"][-1].content)



res3 = agent.invoke(
    {"messages": [{"role": "user", "content": "summarise my history"}]},
    config=config1 # 3
) 

print("\n\n res3")
print(res3["messages"][-1].content)

```

### Create agent with Structured output

```python
from langchain.agents import create_agent
from pydantic import BaseModel

class Answer(BaseModel):
    summary: str
    confidence: float
    name: str



agent = create_agent(
    model="openai:gpt-5.5",
    response_format=Answer
)

text = "NOne"

res = agent.invoke(
    {"messages": [{"role": "user", "content": text}]},
) 

print("\n\nres")
print(res["messages"][-1].content)
# print(res["structured_response"].summary)
# print(res["structured_response"].confidence)


```
