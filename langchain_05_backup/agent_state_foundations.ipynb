
### Foundation: Python dictionaries
A dictionary can represent the current data of an application.

```python
state = {
    "user_id": "user-123",
    "call_count": 0,
}

state["call_count"] += 1
state["status"] = "active"

print(state)
```

### agent needs to track its conversation and other information while working.
State is the data an agent maintains during execution.

```python
state = {
    "messages": [],
    "call_count": 0,
}
```

### Typed state with AgentState
built-in state schema and custom fields.

```python
from langchain.agents import AgentState

class MyState(AgentState):
    user_id: str
    call_count: int
```

### How does LangChain update state?
Middleware returns updates that LangChain applies to the managed state.

```python
return {
    "call_count": state["call_count"] + 1
}
```

### State versus context versus configuration
- state: data maintained by the agent.
- context: per-invocation information available to tools and middleware.
- config: execution configuration, including thread_id.
