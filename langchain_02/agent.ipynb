### create_agent

```python
from langchain.agents import create_agent

agent = create_agent(
    model="openai:gpt-5.6-luna",
    system_prompt="You are my personal assit."
)


res = agent.invoke(
    {
        "messages" : [{
            "role" : "user",
            "content" : "what is agentic ai and what all to learn in it?"
        }]
    }
)
print(res['messages'][-1].content)
```

### create_agent with init_chat_model

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model

chat_model = init_chat_model(
    model="openai:gpt-5.6-luna",
)

agent = create_agent(
    model=chat_model,
    system_prompt="You are my personal assit."
)

res = agent.invoke(
    {
        "messages" : [{
            "role" : "user",
            "content" : "What is cap of India?"
        }]
    }
)
print(res['messages'][-1].content)
```

### init_chat_model - invoke

```python
from langchain.chat_models import init_chat_model

chat_model = init_chat_model(
    model="openai:gpt-5.6-luna",
)
que = "List down the best forest in india?"
res = chat_model.invoke(que)
print(res.content)
```

### init_chat_model - stream

```python
from langchain.chat_models import init_chat_model

chat_model = init_chat_model(
    model="openai:gpt-5.6-luna",
)

que = "List down the best forest in india?"

for chunk in chat_model.stream(que):
    print(chunk.content, end="", flush=True)
```

### init_chat_model - batch and batch_as_completed

```python
from langchain.chat_models import init_chat_model

chat_model = init_chat_model(
    model="openai:gpt-5.6-luna",
)

que = "List down the best forest in india?"

results = chat_model.batch([
    "what is 2+2?",\
    "what is cap of india?"
    "distance between pune to mumbai"
])
for res in results:
    print(res.content)

    
for response in chat_model.batch_as_completed([
    "Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
]):
    print((response))
```
